```python
final_columns = [
    'AppID', 'Name', 'Release date', 'Estimated owners', 'Peak CCU', 
    'Required age', 'Price', 'DiscountDLC count', 'About the game', 
    'Supported languages', 'Full audio languages', 'Reviews', 
    'Header image', 'Website', 'Windows', 'Mac', 'Linux', 
    'Metacritic score', 'User score', 'Positive', 'Negative', 
    'Categories', 'Genres', 'Tags', 'Screenshots'
]


df_filtered = df[final_columns].copy()

print(f"Professional Clean-up Done! New Shape: {df_filtered.shape}")
print("-" * 35)
```

```python
cols_to_fix = ['About the game', 'Genres', 'Tags', 'Categories']
for col in cols_to_fix:
    df_filtered[col] = df_filtered[col].astype(str).replace('nan', '')

df_filtered = df_filtered.dropna(subset=['Name'])

df_filtered.info()
```

```python
df_filtered['metadata'] = (
    df_filtered['Genres'] + ' ' + 
    df_filtered['Tags'] + ' ' + 
    df_filtered['Categories'] + ' ' + 
    df_filtered['About the game']
)

df_filtered['metadata'] = df_filtered['metadata'].str.lower()


print(df_filtered['metadata'].head(1).values[0][:300] + "...") 
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

tfidf = TfidfVectorizer(stop_words='english', max_features=5000)

tfidf_matrix = tfidf.fit_transform(df_filtered['metadata'])

print(f" Matrix Created! Shape: {tfidf_matrix.shape}")
```

```python
def get_recommendations(game_name, tfidf_matrix, df):
    try:
        idx = df[df['Name'].str.contains(game_name, case=False, na=False)].index[0]
    except IndexError:
        return " Game not found. Please try another name."

    cosine_sim = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()

    sim_scores = list(enumerate(cosine_sim))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    sim_scores = sim_scores[1:11]
    game_indices = [i[0] for i in sim_scores]
    scores = [i[1] for i in sim_scores]

    results = df.iloc[game_indices][['Name', 'Genres', 'Tags']].copy()
    results['Match Score'] = scores
    return results

print(" Testing the Recommendation System:")
get_recommendations("Black Dragon Mage", tfidf_matrix, df_filtered)
```

```python
import numpy as np

def evaluate_model_accuracy(game_name, tfidf_matrix, df):
    
    recommendations = get_recommendations(game_name, tfidf_matrix, df)
    
    
    if isinstance(recommendations, str):
        return recommendations
    
    
    accuracy_scores = recommendations['Match Score'].values
    
   
    final_accuracy = np.mean(accuracy_scores) * 100
    
    print(f" Accuracy Report for '{game_name}':")
    print(f"Average Content Similarity: {final_accuracy:.2f}%")
    return final_accuracy


evaluate_model_accuracy("Black Dragon Mage", tfidf_matrix, df_filtered)
```

```python
import random

def calculate_global_accuracy(tfidf_matrix, df, n_tests=100):
    all_scores = []
    
    random_games = random.sample(list(df['Name']), n_tests)
    
    for game in random_games:
        try:
            
            recs = get_recommendations(game, tfidf_matrix, df)
            if not isinstance(recs, str):
                all_scores.append(np.mean(recs['Match Score']))
        except:
            continue
            
    global_avg = np.mean(all_scores) * 100
    print(f" Global Model Accuracy (Tested on {len(all_scores)} games):")
    print(f"Final Score: {global_avg:.2f}%")
    return global_avg


calculate_global_accuracy(tfidf_matrix, df_filtered)
```
